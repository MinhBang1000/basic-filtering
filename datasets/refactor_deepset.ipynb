```python
from datasets import concatenate_datasets

print("Loading deepset/prompt-injections dataset...\n")
ds = load_dataset("deepset/prompt-injections")
# Concat the two splits together

ds_combined = concatenate_datasets([ds["train"], ds["test"]])
ds_combined
```

```python
def add_columns(example):
    example["original_label"] = example["label"]
    example["label"] = 1
    example["label_note"] = "Suspicious Data"
    return example

ds_combined = ds_combined.map(add_columns)
print(ds_combined)
```

```python
import pandas as pd
import numpy as np

df = ds_combined.to_pandas()
df.head()
df.to_json("deepset_prompt_injections_refactored.jsonl", orient="records", lines=True)
```
