# Extract Data

```python
import pandas as pd
```

```python
from pathlib import Path

# 1. Define base directory
DS_DIR = Path("../injectagent")

# 2. Extract paths for ALL files in that folder
# This creates a dictionary: {'tools': '../injectagent/tools.json', ...}
file_paths = {f.stem: f for f in DS_DIR.iterdir() if f.is_file()}

# 3. Print the results to verify
for name, path in file_paths.items():
    print(f"{name: <30} -> {path}")
```

```python
# Read those files:
dataframes = {}

# Loop through all files in the directory
for file_path in DS_DIR.glob("*.json*"):
    name = file_path.stem  # This gets the filename without extension
    
    try:
        if file_path.suffix == ".jsonl":
            # JSONL files require lines=True
            dataframes[name] = pd.read_json(file_path, lines=True)
        else:
            # Standard JSON files
            dataframes[name] = pd.read_json(file_path)
            
        print(f"✅ Loaded: {name} ({len(dataframes[name])} rows)")
    except Exception as e:
        print(f"❌ Failed to load {name}: {e}")
```

```python
# Only for the attacker_simulated_responses file
# Use this for the specific file causing the error
sim_responses = pd.read_json(DS_DIR / "attacker_simulated_responses.json", typ='series')
# Accessing it
print("attackr_simulated_responses.json")
print(sim_responses.head())
print()

# Print columns for all remaining loaded files
for name, df in dataframes.items():
    print(f"{name} columns: {list(df.columns)}")
```

```python
# Extract numbers of tools we have to use
target_files = ["user_cases", "attacker_cases_ds", "attacker_cases_dh"]
for name, df in dataframes.items():
    if name in target_files:
        print(f"{name} data:")
        for index, row in df.iterrows():
            print(row.to_dict())
        print()
```
